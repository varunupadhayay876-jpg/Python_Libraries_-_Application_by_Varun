```python
df["Bonus"] = 5
print(df)
```

```python
df["FinalMarks"] = df["Marks"] + df["Bonus"]
print(df)
```

```python
df["Result"] = df["Marks"].apply(lambda x: "Pass" if x>=40 else "Fail")
print(df["Result"])
```

```python
def grade(x):
    if x>=90: return "A+"
    if x>=80: return "A"
    if x>=70: return "B"
    if x>=60: return "C"
    return "D"

df["Grade"] = df["Marks"].apply(grade)
print(df[["Name","Marks","Grade"]])
```

```python
print(df.sort_values("Marks", ascending=False))
```

```python
df["Rank"] = df["Marks"].rank(ascending=False).astype(int)
print(df[["Name","Marks","Rank"]])
```

```python

import pandas as pd

data = {
    "RollNo": [101, 102, 103, 104, 105, 106, 107, 108],
    "Name": [
        "Amit", "Neha", "Rahul", "Priya",
        "Rohan", "Kavya", "Arjun", "Simran"
    ],
    "Department": [
        "BCA", "BCA", "MCA", "BCA",
        "MCA", "MCA", "BCA", "MCA"
    ],
    "Age": [20, 21, 22, 20, 23, 22, 21, 24],
    "Marks": [85, 92, 67, 78, 88, 95, 55, 73]
}

df = pd.DataFrame(data)

print(df)
```

```python

# Q1. Display complete DataFrame
print(df)

# Q2. First 3 records
print(df.head(3))

# Q3. Last 2 records
print(df.tail(2))

# Q4. Number of rows and columns
print(df.shape)

# Q5. All column names
print(df.columns)

# Q6. Data types of all columns
print(df.dtypes)

# Q7. Only Name column
print(df["Name"])

# Q8. Name and Marks columns
print(df[["Name", "Marks"]])
```

```python

# Q9. Students scoring more than 80
print(df[df["Marks"] > 80])

# Q10. Students scoring less than 70
print(df[df["Marks"] < 70])

# Q11. Marks between 70 and 90 (inclusive)
print(df[df["Marks"].between(70, 90)])

# Q12. Students belonging to BCA department
print(df[df["Department"] == "BCA"])

# Q13. MCA students scoring more than 80
print(df[
    (df["Department"] == "MCA") &
    (df["Marks"] > 80)
])

# Q14. Students whose age is greater than 21
print(df[df["Age"] > 21])
```

```python

# Q15. Average marks
print(df["Marks"].mean())

# Q16. Maximum marks
print(df["Marks"].max())

# Q17. Minimum marks
print(df["Marks"].min())

# Q18. Median marks
print(df["Marks"].median())

# Q19. Number of students scoring above 75
print((df["Marks"] > 75).sum())

# Q20. Average marks of BCA students
print(df[df["Department"] == "BCA"]["Marks"].mean())

# Q21. Average marks of MCA students
print(df[df["Department"] == "MCA"]["Marks"].mean())

# Q22. Highest-scoring student
print(df[df["Marks"] == df["Marks"].max()])
```

```python

# Q23. Add Bonus column
df["Bonus"] = 5

# Q24. Create FinalMarks column
df["FinalMarks"] = df["Marks"] + df["Bonus"]

# Q25. Create Result column
df["Result"] = df["Marks"].apply(
    lambda x: "Pass" if x >= 40 else "Fail"
)

# Q26. Create Grade column
def calculate_grade(marks):
    if marks >= 90:
        return "A+"
    elif marks >= 80:
        return "A"
    elif marks >= 70:
        return "B"
    elif marks >= 60:
        return "C"
    else:
        return "D"

df["Grade"] = df["Marks"].apply(calculate_grade)

print(df)
```

```python

# Q27. Sort marks from highest to lowest
print(df.sort_values(by="Marks", ascending=False))

# Q28. Display top 3 students
print(df.nlargest(3, "Marks"))

# Q29. Add Rank column
df["Rank"] = df["Marks"].rank(
    ascending=False,
    method="min"
).astype(int)

print(df)
```

```python

# Q30. Number of students in each department
print(df.groupby("Department")["Name"].count())

# Q31. Average marks of each department
print(df.groupby("Department")["Marks"].mean())

# Q32. Maximum marks in each department
print(df.groupby("Department")["Marks"].max())

# Q33. Minimum marks in each department
print(df.groupby("Department")["Marks"].min())

# Combined analysis (all four together)
print(
    df.groupby("Department")["Marks"].agg(
        ["count", "mean", "max", "min"]
    )
)
```

```python

# Challenge Question: Final Merit List

merit_list = df[
    ["Rank", "RollNo", "Name", "Department",
     "Marks", "Grade", "Result"]
].sort_values(by="Marks", ascending=False)

print(merit_list)

# Export to Excel
merit_list.to_excel(
    "student_merit_list.xlsx",
    index=False
)

print("Merit list exported successfully!")
```

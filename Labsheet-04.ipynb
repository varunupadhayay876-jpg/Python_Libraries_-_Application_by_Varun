```python
# Q2. Linear Regression using TensorFlow
import tensorflow as tf

x = tf.constant([1., 2., 3.])
y = tf.constant([2., 4., 6.])

model = tf.keras.Sequential([
    tf.keras.Input(shape=(1,)),
    tf.keras.layers.Dense(1)
])

model.compile(optimizer='sgd', loss='mse')
model.fit(x, y, epochs=10, verbose=0)

print("Model trained successfully")
print(model.predict(tf.constant([[5.]]), verbose=0))
```

```python
# Q4. Calculate gradient
import tensorflow as tf

x = tf.Variable(5.0)

with tf.GradientTape() as tape:
    y = x ** 2

print(tape.gradient(y, x).numpy())
```

```python
# Q5. Custom loss function
import tensorflow as tf

def my_loss(y_true, y_pred):
    return tf.reduce_mean((y_true - y_pred) ** 2)

model = tf.keras.Sequential([
    tf.keras.Input(shape=(1,)),
    tf.keras.layers.Dense(1)
])

model.compile(optimizer='adam', loss=my_loss)
print("Custom loss applied")
```

```python
# Q7. Sine function
import tensorflow as tf
import numpy as np

x = np.linspace(-3, 3, 50)
y = np.sin(x)

model = tf.keras.Sequential([
    tf.keras.Input(shape=(1,)),
    tf.keras.layers.Dense(10, activation='relu'),
    tf.keras.layers.Dense(1)
])

model.compile(optimizer='adam', loss='mse')
model.fit(x, y, epochs=10, verbose=0)

print("Sine model trained")
```

```python
# Q8. Binary classification
import tensorflow as tf

model = tf.keras.Sequential([
    tf.keras.Input(shape=(2,)),
    tf.keras.layers.Dense(8, activation='relu'),
    tf.keras.layers.Dense(1, activation='sigmoid')
])

model.compile(optimizer='adam',
              loss='binary_crossentropy',
              metrics=['accuracy'])

print("Binary model created")
```

```python
# Q11. EarlyStopping and ModelCheckpoint
import tensorflow as tf
import numpy as np

x = np.array([1., 2., 3., 4.])
y = np.array([2., 4., 6., 8.])

model = tf.keras.Sequential([
    tf.keras.Input(shape=(1,)),
    tf.keras.layers.Dense(1)
])

model.compile(optimizer='adam', loss='mse')

callbacks = [
    tf.keras.callbacks.EarlyStopping(
        monitor='val_loss', patience=2
    ),
    tf.keras.callbacks.ModelCheckpoint(
        'best.keras', monitor='val_loss',
        save_best_only=True
    )
]

model.fit(x, y, epochs=10,
          validation_split=0.25,
          callbacks=callbacks)

print("Training completed")
```

```python
# Q12. Dropout layer
import tensorflow as tf

model = tf.keras.Sequential([
    tf.keras.Input(shape=(10,)),
    tf.keras.layers.Dense(16, activation='relu'),
    tf.keras.layers.Dropout(0.5),
    tf.keras.layers.Dense(1)
])

model.summary()
```

```python
# Q13. Functional API
import tensorflow as tf

a = tf.keras.Input(shape=(2,))
b = tf.keras.Input(shape=(2,))

x = tf.keras.layers.Concatenate()([a, b])
y1 = tf.keras.layers.Dense(1)(x)
y2 = tf.keras.layers.Dense(1)(x)

model = tf.keras.Model([a, b], [y1, y2])
model.summary()
```

```python
# Q14. Hyperparameter tuning
import tensorflow as tf
import keras_tuner as kt

def build_model(hp):
    model = tf.keras.Sequential([
        tf.keras.Input(shape=(2,)),
        tf.keras.layers.Dense(
            hp.Choice('units', [8, 16, 32]),
            activation='relu'),
        tf.keras.layers.Dense(1)
    ])
    model.compile(optimizer='adam', loss='mse')
    return model

tuner = kt.RandomSearch(
    build_model, objective='val_loss',
    max_trials=3, directory='tune'
)

print("Tuner created")
```

```python
# Q15. Plot training curves
import tensorflow as tf
import matplotlib.pyplot as plt
import numpy as np

x = np.random.rand(100, 2)
y = np.random.randint(0, 2, 100)

model = tf.keras.Sequential([
    tf.keras.Input(shape=(2,)),
    tf.keras.layers.Dense(8, activation='relu'),
    tf.keras.layers.Dense(1, activation='sigmoid')
])

model.compile(optimizer='adam',
              loss='binary_crossentropy',
              metrics=['accuracy'])

h = model.fit(x, y, epochs=5, validation_split=0.2)

plt.plot(h.history['loss'], label='Training')
plt.plot(h.history['val_loss'], label='Validation')
plt.legend()
plt.show()
```
